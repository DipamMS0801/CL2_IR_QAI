```python
import networkx as nx

# Directed Graph
G = nx.DiGraph()

edges = [
    (0,1),
    (1,2),
    (2,0),
    (2,3),
    (3,1)
]

G.add_edges_from(edges)

# PageRank
pr = nx.pagerank(
    G,
    alpha=0.85
)

print("PageRank Scores:\n")

for node, score in pr.items():
    print(f"Node {node}: {score:.4f}")
```
