```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Ensure NLTK resources are available
nltk.download('stopwords')
nltk.download('punkt') # Added to download the 'punkt' tokenizer data
nltk.download('punkt_tab') # Added to download the 'punkt_tab' resource as specified by the error

# 1. Create a sample file as instructed
with open('file.txt', 'w', encoding='utf8') as f:
    f.write("This is the first word.\n")
    f.write("This is the second text, Hello! How are you?\n")
    f.write("This is the third, this is it now.\n")

# 2. Read and process the file
with open('file.txt', 'r', encoding='utf8') as file:
    lines = file.readlines()

stop_words = set(stopwords.words('english'))
inverted_index = {}

# 3. Build the Inverted Index
for line_num, line in enumerate(lines, 1):
    # Remove punctuation and convert to lowercase
    clean_line = "".join([char for char in line.lower() if char.isalnum() or char.isspace()])

    # Tokenize and remove stop words
    tokens = word_tokenize(clean_line)
    tokens_without_sw = [word for word in tokens if word not in stop_words]

    for word in tokens_without_sw:
        if word not in inverted_index:
            inverted_index[word] = []
        if line_num not in inverted_index[word]:
            inverted_index[word].append(line_num)

print("Inverted Index:\n", inverted_index)
```
