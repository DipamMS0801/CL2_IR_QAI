```python
# Install pgmpy if it's not already installed
!pip install pgmpy pandas numpy

import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

# Load dataset
df = pd.read_csv("/content/heart.csv")

print(df.head())

# Define Network Structure
model = DiscreteBayesianNetwork([
    ('cp', 'target'),
    ('thal', 'target'),
    ('ca', 'target'),
    ('oldpeak', 'target')
])

# Learn Parameters
print('Learning CPD using Maximum likelihood estimators...')
model.fit(df) # Removed estimator argument to let it default to DiscreteMLE()

print("Model Learned Successfully")

# Inference
infer = VariableElimination(model)

q = infer.query(
    variables=['target'],
    evidence={'cp': 2}
)

print(q)

# Computing the Probability of Heart Disease given chest pain type
print('\n2. Probability of HeartDisease given evidence= cp')
q2 = infer.query(variables=['target'], evidence={'cp': 2})
print(q2)
```
