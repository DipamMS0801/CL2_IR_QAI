```python

import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork

# Load dataset
df = pd.read_csv("C:\\Users\\Ai\\OneDrive\\Desktop\\CL2_IR_QAI\\IR_Practicals\\P3_heart.csv")

print(df.head())

# Define Network Structure
model = DiscreteBayesianNetwork([
    ('cp', 'target'),
    ('thal', 'target'),
    ('ca', 'target'),
    ('oldpeak', 'target')
])

# Estimate the requested conditional probabilities
print('Learning conditional probabilities...')
target_states = sorted(df['target'].dropna().unique())
target_counts = df.groupby('cp', observed=False)['target'].value_counts()
cpd = target_counts.unstack(fill_value=0).reindex(columns=target_states, fill_value=0)
cpd = cpd.div(cpd.sum(axis=1), axis=0)

print("Model Learned Successfully")

# Inference
infer = cpd.loc[2]

q = infer

print(q)

# Computing the Probability of Heart Disease given chest pain type
print('\n2. Probability of HeartDisease given evidence= cp')
q2 = infer
print(q2)
```
