```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download required NLTK data (run these once)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Added to resolve LookupError

def preprocess_text(document):
    # 1. Lowercase the text to ensure uniformity
    document = document.lower()

    # 2. Tokenization: Break the text into individual words
    words = word_tokenize(document)

    # 3. Stop Word Removal: Filter out common words (e.g., 'the', 'is', 'in')
    stop_words_set = set(stopwords.words('english'))
    filtered_words = [word for word in words if word not in stop_words_set and word.isalnum()]

    # 4. Stemming: Reduce words to their root/base form
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(word) for word in filtered_words]

    return stemmed_words

# Sample text document
sample_document = """
    Information retrieval is the process of obtaining information system resources
    that are relevant to an information need from a collection of those resources.
    Web mining applies data mining techniques to discover patterns from the World Wide Web.
"""

print("Original Document:\n", sample_document.strip())

# Process the text
processed_text = preprocess_text(sample_document)

print("\nProcessed Document (Tokens):\n", processed_text)
```
